```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

# Normal Equation: θ_hat = (X⊺X)^−1 X⊺ y
θ = containing both bias and weigths parameters

To find parameter theta(θ) that minimize the cost function, closer form solution.

Mathematically, to used the equation and directly find the result.


```python
X = 5 * np.random.rand(100, 1) # feature
y = 4 * X + np.random.randn(100, 1) #actual value

# now compute theta using Normal Equation
X_b = np.c_[np.ones(X.shape), X]  # X_b -> bias augmented feature matrix

theta_hat = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)  

# OR
# theta_hat = np.linalg.inv(np.dot(X_b.T , X_b)).dot(np.dot(X_b.T, y))

# predication: y_hat = Xb.theta_hat

y_predict = X_b.dot(theta_hat)
print(f"Precicated output =\n {y_predict[:5]}")
print(f"\nActual value =\n {y[:5]}")
# print(X_b[:4])
```

# SVD approach  

```python
# theta, residual, rank, s  = np.linalg.lstsq(X_b, y, rcond=1e-16)
# OR
theta_svd = np.linalg.pinv(X_b).dot(y)
print(f"SVD approach =\n {theta_svd}")
print(f"\nMathematical/Normal Equation =\n {theta_hat}")
```

# Testing for new data

```python
X_new = np.array([1,3,5]).reshape(-1, 1)
X_new_b = np.c_[np.ones(X_new.shape), X_new]

# predication
y_hat = X_new_b.dot(theta_hat)
y_hat
```
