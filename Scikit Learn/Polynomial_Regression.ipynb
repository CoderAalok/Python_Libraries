```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
m = 100
X = 8 * np.random.rand(m, 1) - 4
# add power on feature as new feature
y = 0.5 * X**2 + X + 2 + np.random.randn(m, 1)  # quadratic equation : y = ax2 + bx + c

poly_features = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly_features.fit_transform(X)

model = LinearRegression()
# fit linear model
model.fit(X_poly, y)

# predication
y_hat = model.predict(X_poly)

# sorting
idx = np.argsort(X[:, 0])
X_sort = X[idx]
y_hat_sort = y_hat[idx]

# plot
fig, ax = plt.subplots(figsize=(9,6))

ax.scatter(X, y)
ax.plot(X_sort, y_hat_sort, color='r', linewidth=2, label='Predication')

ax.set_xlabel('X')
ax.set_ylabel('y')
ax.legend()

plt.show()
```
